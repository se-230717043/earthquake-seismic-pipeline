```python
import sqlite3

conn = sqlite3.connect('../data/earthquake.db')

print("--- Data Profiling Summary ---")
print("Raw Events Count:", conn.execute("SELECT COUNT(*) FROM earthquake_events").fetchone()[0])
print("Clean Events Count:", conn.execute("SELECT COUNT(*) FROM earthquake_events_clean").fetchone()[0])
print("Distinct Places:", conn.execute("SELECT COUNT(DISTINCT place) FROM earthquake_events_clean").fetchone()[0])

rows = conn.execute("SELECT magnitude, place FROM earthquake_events_clean LIMIT 5").fetchall()
print("\nSample Cleaned Data:")
for row in rows:
    print(row)

conn.close()
```
